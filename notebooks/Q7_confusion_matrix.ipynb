# Q7 — Best Model Test Accuracy + Confusion Matrix

```python
import sys
sys.path.append('..')
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pickle
from utils import load_fashion_mnist, compute_accuracy, FASHION_LABELS
```

```python
X_train, y_train, X_test, y_test = load_fashion_mnist()
```

```python
with open('../models/best_model.pkl', 'rb') as f:
    model = pickle.load(f)
print(f'architecture: {model.layer_sizes}')
print(f'activation: {model.activation}')
```

```python
y_pred = model.predict(X_test)
test_acc = compute_accuracy(y_test, y_pred)
print(f'Test Accuracy: {test_acc:.2%}')
```

```python
# confusion matrix
cm = np.zeros((10, 10), dtype=int)
for t, p in zip(y_test, y_pred):
    cm[t][p] += 1

cm_norm = cm.astype(float) / cm.sum(axis=1, keepdims=True)

print('Per-class accuracy:')
for i, label in enumerate(FASHION_LABELS):
    print(f'  {label:15s}: {cm_norm[i,i]:.2%}')
```

```python
# creative dual confusion matrix
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7))

sns.heatmap(cm, annot=True, fmt='d', cmap=sns.color_palette('rocket_r', as_cmap=True),
            ax=ax1, xticklabels=FASHION_LABELS, yticklabels=FASHION_LABELS,
            linewidths=0.5, linecolor='gray')
ax1.set_title('Confusion Matrix (Raw Counts)', fontsize=13, pad=12)
ax1.set_xlabel('Predicted'); ax1.set_ylabel('True')
plt.setp(ax1.get_xticklabels(), rotation=45, ha='right')

sns.heatmap(cm_norm, annot=True, fmt='.1%', cmap='YlOrRd', ax=ax2,
            xticklabels=FASHION_LABELS, yticklabels=FASHION_LABELS,
            linewidths=0.5, linecolor='gray', vmin=0, vmax=1)
ax2.set_title('Confusion Matrix (Normalized)', fontsize=13, pad=12)
ax2.set_xlabel('Predicted'); ax2.set_ylabel('True')
plt.setp(ax2.get_xticklabels(), rotation=45, ha='right')

plt.suptitle(f'Fashion-MNIST — Test Accuracy: {test_acc:.2%}', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

### Observations
- **Shirt (70.0%)** is hardest — confused with T-shirt/top (12.6%) and Coat (7.4%)
- **Trouser (97.3%), Sneaker (97.0%), Bag (98.0%)** are easiest — visually distinct shapes
- **T-shirt vs Shirt** is the most confused pair — both are upper body garments
- **Sneaker vs Ankle boot** also confused (5.3%) — similar footwear shapes
